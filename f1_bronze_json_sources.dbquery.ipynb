```sql
USE CATALOG f1_project;

-- ============================================================
-- F1 PROJECT - BRONZE LAYER
-- Raw ingestion from JSON files
-- ============================================================

-- ============================================================
-- DRIVERS
-- ============================================================

CREATE TABLE IF NOT EXISTS f1_project.bronze.drivers_raw (
  driver_id STRING,
  driver_ref STRING,
  number STRING,
  code STRING,
  forename STRING,
  surname STRING,
  dob STRING,
  nationality STRING,
  _ingested_at TIMESTAMP,
  _source_file STRING
);

COPY INTO f1_project.bronze.drivers_raw
FROM (
  SELECT
    CAST(driver_id AS STRING) AS driver_id,
    CAST(driver_ref AS STRING) AS driver_ref,
    CAST(number AS STRING) AS number,
    CAST(code AS STRING) AS code,
    CAST(forename AS STRING) AS forename,
    CAST(surname AS STRING) AS surname,
    CAST(dob AS STRING) AS dob,
    CAST(nationality AS STRING) AS nationality,
    current_timestamp() AS _ingested_at,
    _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/drivers.json'
)
FILEFORMAT = JSON;


-- ============================================================
-- CONSTRUCTORS
-- ============================================================

CREATE TABLE IF NOT EXISTS f1_project.bronze.constructors_raw (
  constructor_id STRING,
  constructor_ref STRING,
  name STRING,
  nationality STRING,
  _ingested_at TIMESTAMP,
  _source_file STRING
);

COPY INTO f1_project.bronze.constructors_raw
FROM (
  SELECT
    CAST(constructor_id AS STRING) AS constructor_id,
    CAST(constructor_ref AS STRING) AS constructor_ref,
    CAST(name AS STRING) AS name,
    CAST(nationality AS STRING) AS nationality,
    current_timestamp() AS _ingested_at,
    _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/constructors.json'
)
FILEFORMAT = JSON;


-- ============================================================
-- PIT STOPS
-- ============================================================

CREATE TABLE IF NOT EXISTS f1_project.bronze.pit_stops_raw (
  race_id STRING,
  driver_id STRING,
  stop STRING,
  lap STRING,
  time STRING,
  duration STRING,
  _ingested_at TIMESTAMP,
  _source_file STRING
);

COPY INTO f1_project.bronze.pit_stops_raw
FROM (
  SELECT
    CAST(race_id AS STRING) AS race_id,
    CAST(driver_id AS STRING) AS driver_id,
    CAST(stop AS STRING) AS stop,
    CAST(lap AS STRING) AS lap,
    CAST(time AS STRING) AS time,
    CAST(duration AS STRING) AS duration,
    current_timestamp() AS _ingested_at,
    _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/pit_stops.json'
)
FILEFORMAT = JSON;


-- ============================================================
-- SANITY CHECK
-- ============================================================

SELECT
  'drivers' AS tbl,
  COUNT(*) AS cnt
FROM f1_project.bronze.drivers_raw
UNION ALL
SELECT
  'constructors' AS tbl,
  COUNT(*) AS cnt
FROM f1_project.bronze.constructors_raw
UNION ALL
SELECT
  'pit_stops' AS tbl,
  COUNT(*) AS cnt
FROM f1_project.bronze.pit_stops_raw;


-- ============================================================
-- SCHEMA CHECK
-- ============================================================
select * from f1_project.bronze.drivers_raw;

select * from f1_project.bronze.constructors_raw;

select * from f1_project.bronze.pit_stops_raw;



```

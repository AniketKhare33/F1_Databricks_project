```sql
USE CATALOG f1_project;

-- ============================================================
-- CREATE TABLE
-- ============================================================

CREATE TABLE IF NOT EXISTS f1_project.bronze.qualifying_raw (
  qualify_id STRING,
  race_id STRING,
  driver_id STRING,
  constructor_id STRING,
  number STRING,
  position STRING,
  q1 STRING,
  q2 STRING,
  q3 STRING,
  _ingested_at TIMESTAMP,
  _source_file STRING
);


-- ============================================================
-- LOAD JSON
-- Explicitly cast all business columns to STRING
-- ============================================================

COPY INTO f1_project.bronze.qualifying_raw
FROM (
  SELECT
    CAST(qualify_id AS STRING) AS qualify_id,
    CAST(race_id AS STRING) AS race_id,
    CAST(driver_id AS STRING) AS driver_id,
    CAST(constructor_id AS STRING) AS constructor_id,
    CAST(number AS STRING) AS number,
    CAST(position AS STRING) AS position,
    CAST(q1 AS STRING) AS q1,
    CAST(q2 AS STRING) AS q2,
    CAST(q3 AS STRING) AS q3,
    current_timestamp() AS _ingested_at,
    _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/qualifying.json'
)
FILEFORMAT = JSON
FORMAT_OPTIONS ('multiLine' = 'true');


-- ============================================================
-- SANITY CHECK
-- ============================================================

SELECT COUNT(*) AS cnt
FROM f1_project.bronze.qualifying_raw;

SELECT *
FROM f1_project.bronze.qualifying_raw
LIMIT 5;



-- ============================================================
-- SCHEMA CHECK
-- ============================================================

DESCRIBE TABLE f1_project.bronze.qualifying_raw;
```

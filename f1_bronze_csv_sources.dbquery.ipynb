```sql
USE CATALOG f1_project;


create table if not exists f1_project.bronze.circuits_raw (
  circuit_id STRING,
  circuit_ref STRING, 
  name STRING, 
  location STRING,
  country STRING, 
  lat STRING, 
  lng STRING, 
  alt STRING,
  _ingested_at TIMESTAMP, 
  _source_file STRING
);


COPY INTO f1_project.bronze.circuits_raw
FROM (
  SELECT *, current_timestamp() AS _ingested_at, _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/circuits.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');


CREATE TABLE IF NOT EXISTS f1_project.bronze.races_raw (
  race_id STRING, year STRING, round STRING, circuit_id STRING,
  name STRING, date STRING, time STRING,
  _ingested_at TIMESTAMP, _source_file STRING
);

COPY INTO f1_project.bronze.races_raw
FROM (
  SELECT *, current_timestamp() AS _ingested_at, _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/races.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');


-- results
CREATE TABLE IF NOT EXISTS f1_project.bronze.results_raw (
  result_id STRING, race_id STRING, driver_id STRING, constructor_id STRING,
  grid STRING, position STRING, points STRING, laps STRING, time STRING,
  fastest_lap STRING, rank STRING, fastest_lap_time STRING, status STRING,
  _ingested_at TIMESTAMP, _source_file STRING
);

COPY INTO f1_project.bronze.results_raw
FROM (
  SELECT *, current_timestamp() AS _ingested_at, _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/results.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');

-- lap_times
CREATE TABLE IF NOT EXISTS f1_project.bronze.lap_times_raw (
  race_id STRING, driver_id STRING, lap STRING, position STRING,
  time STRING, milliseconds STRING,
  _ingested_at TIMESTAMP, _source_file STRING
);


COPY INTO f1_project.bronze.lap_times_raw
FROM (
  SELECT *, current_timestamp() AS _ingested_at, _metadata.file_path AS _source_file
  FROM '/Volumes/f1_project/bronze/landing/lap_times.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');


SELECT 'circuits' AS tbl, count(*) AS cnt FROM f1_project.bronze.circuits_raw
UNION ALL SELECT 'races', count(*) FROM f1_project.bronze.races_raw
UNION ALL SELECT 'results', count(*) FROM f1_project.bronze.results_raw
UNION ALL SELECT 'lap_times', count(*) FROM f1_project.bronze.lap_times_raw;



```
